## Q3. Employee Roles | Beginner
Create a parent class Employee that stores a name and employee ID and provides display_details(). Derive
Developer (programming language) and Manager (team size). Use super() to initialize inherited attributes.
Override display_details() in each child class to include both shared and role-specific information. Create and test
one object of each type.

```python
class Employee:
    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id

    def display_details(self):
        print(f"Employee Name: {self.name}, Employee ID: {self.employee_id}")

class Manager(Employee):
    def __init__(self, name, employee_id, department):
        super(). __init__(name, employee_id)  # Call the constructor of the base class
        self.department = department

    def display_details(self):
        print(f"Manager name: {self.name}, Employee ID: {self.employee_id}, Department: {self.department}")

class Developer(Employee):
    def __init__(self, name, employee_id, programming_language):
        super(). __init__(name, employee_id)  # Call the constructor of the base class
        self.programming_language = programming_language

    def display_details(self):
        print(f"Developer name: {self.name}, Employee ID: {self.employee_id}, Programming Language: {self.programming_language}")

# Creating objects of Manager and Developer classes
manager1 = Manager("Uzair", 102, "IT")
developer1 = Developer("Ahmed", 103, "Python")

# Displaying details of Manager and Developer
manager1.display_details()
developer1.display_details()
```

## Q4. Vehicle Rental System | Intermediate
Create a base class Vehicle with brand, model, and daily rental rate. Implement calculate_rent(days). Derive Car,
which adds a fixed insurance fee of Rs. 1,000 per rental, and Bike, which applies a 10% discount to the base
rental amount. Override calculate_rent() and reuse the parent implementation with super(). Reject rental
durations of zero or fewer days. Test both vehicle types for a three-day rental.

```python
class Vehicle:
    def __init__(self, brand, model, daily_rental_price):
        self.brand = brand
        self.model = model
        self.daily_rental_price = daily_rental_price

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("Rental duration must be a positive number of days.")
        return self.daily_rental_price * days

class Car(Vehicle):
    def __init__(self, brand, model, daily_rental_price):
        super(). __init__(brand, model, daily_rental_price)
        self.insurance_fee = 1000

    def calculate_rent(self, days):
        base_rent = super().calculate_rent(days)
        total_rent = base_rent + self.insurance_fee
        return total_rent

class Bike(Vehicle):
    def __init__(self, brand, model, daily_rental_price):
        super(). __init__(brand, model, daily_rental_price)

    def calculate_rent(self, days):
        base_rent = super().calculate_rent(days)
        discount = base_rent * 0.10
        total_rent = base_rent - discount
        return total_rent
```

```python
# Creating objects of Car and Bike classes
car1 = Car("Toyota", "Corolla", 2000)
bike1 = Bike("Suzuki", "GCR", 1000)

# Calculating rent for 5 days
print("\n- - CAR RENTAL CALCULATION - -")
print(f"The total rental cost for the car is: {car1.calculate_rent(5)}")
print(f"The total rental cost for the bike is: {bike1.calculate_rent(5)}")

print(car1.calculate_rent(3))
print(bike1.calculate_rent(3))
```

Q22. Set Operations

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

Find union, intersection, a - b, and symmetric difference.

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}

union = a.union(b)
print("The Union are: ", union)

intersection = a.intersection(b)
print("The Intersection are: ", intersection)

difference = a.difference(b)
print("The differences are: ", difference)

symmetrical_differnece = a.symmetric_difference(b)
print("The symmetrical difference are: ", symmetrical_differnece)
```

Q30. Customer Segmentation

`website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}`

Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

# All three platforms
all_three = website & mobile & store

# At least two platforms (helper set)
at_least_two = (website & mobile) | (website & store) | (mobile & store)

# Exactly two platforms
exactly_two = at_least_two - all_three

# Exactly one platform
exactly_one = (website | mobile | store) - at_least_two

print("Exactly One Platform: ", sorted(exactly_one))
print("Exactly Two Platform: ", sorted(exactly_two))
print("At least two Platform: ", sorted(at_least_two))
print("All three Platform: ", sorted(all_three))
```

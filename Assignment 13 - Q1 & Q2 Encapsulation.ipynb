# Q1. Student Result Manager | Beginner
Create a Student class with a public name, protected _marks, and private __grade attributes. Initialize the
student through a constructor. Implement set_marks(marks), which accepts only values from 0 to 100, along with
get_marks() and get_grade(). Assign grade A for 80 to 100, B for 60 to 79, C for 40 to 59, and F below 40.
Demonstrate both a valid and an invalid marks update. Extension: Attempt direct access to __grade and explain
the result.

```python
class Student:
    def __init__(self, name, marks): # Constructor method to initialize the Student object
        self.name = name # Public attribute
        self._marks = marks # Protected attribute
        self.__grade = "F" # Private attribute
        self.set_marks(marks) # Call the set_marks method to initialize the grade based on marks

    def set_marks(self, marks):
        if marks >= 0 and marks <=100:
            #self._marks = marks

            if marks >=80:
                self.__grade = 'A'
            elif marks >= 60:
                self.__grade = 'B'
            elif marks >= 40:
                self.__grade = 'C'
            else:
                self.__grade = 'F'
            print("Marks updated successfully.")
        else:
            print("Invalid marks. Please enter a value between 0 and 100.")

    def get_marks(self):
        return self._marks
    def get_grade(self):
        return self.__grade
```

# Q2. Secure Bank Account | Intermediate
Create a BankAccount class with a public account holder name, private __balance, and private __pin. Implement
deposit(amount), withdraw(amount, pin), and get_balance(pin). Reject zero or negative deposits, withdrawals
exceeding the available balance, and incorrect PINs. Demonstrate a successful deposit, a successful withdrawal,
and two rejected operations. Extension: Explain why double underscores do not provide true security.

```python
class BankAccount:
    def __init__(self, name, balance, pin):
        self.name = name # Public attribute
        self.__balance = balance # Private attribute
        self.__pin = pin # Private attribute

    def deposit(self, amount):
        if amount <= 0:
            print(f"Rejected: The Deposit amount must be greater than zero.")
        else:
            self.__balance += amount
            print(f"Deposited: {amount}. New balance: {self.__balance}")

    def withdrawl(self, pin, amount):
        if pin != self.__pin:
            print(f"Rejected: Invalid Pin.")
        elif amount <= 0:
            print(f"Rejected: The withdrawl amount must be greater than zero.")
        elif amount > self.__balance:
            print(f"Rejected: Insufficient funds.")
        else:
            self.__balance -= amount
            print(f"Withdrawl {amount} successful. New balance: {self.__balance}")

    def get_balance(self, pin):
        if pin != self.__pin:
            print("Rejected: Invalid Pin.")
        else:
            print(f"Current balance is: {self.__balance}")
```

```python
# Crating BankAccount objects
account1 = BankAccount("Uzair", 5000, 1234)

# Display account information
print(f"Account Holder: {account1.name}")

# Successful deposit
print("\n- - Successful Deposit - -")
account1.deposit(1000)

# Successful withdrawl
print("\n- - Successful Withdrawl - -")
account1.withdrawl(1234, 2000)

# Rejected withdrawl due to insufficient funds
print("\n- - Rejected Withdrawl - -")
account1.withdrawl(1234, 5000)

# Rejected withdrawl due to invalid pin
print("\n - Rejected Withdrawl - -")
account1.withdrawl(4321, 1000)

# Display current balance with valid pin
print("\n- - Display Current Balance - -")
account1.get_balance(1234)

## Python performs Name Mangling. It internally changes the attribute name from __balance to _BankAccount__balance. Therefore, double 
#underscores discourage accidental access; they do not encrypt or securely hide the data.Real banking systems require proper 
#authentication, encryption, secure PIN hashing and access controls.
```

Q8. Flatten into [10, 20, 30, 40, 50, 60] using list operations.
data = [[10, 20], [30, 40], [50, 60]]

```python
data = [[10,20],[30,40],[50,60]]

Flattened_list = []

for sublist in data:
    Flattened_list.extend(sublist)

print(Flattened_list)
```

Q9. Extract Data from a Matrix
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
Extract main diagonal, secondary diagonal, and middle column; calculate the sum of each.

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diagonal = [matrix[i][i] for i in range(3)]

secondary_diagonal = [matrix[i][2-i] for i in range(3)]

middle_column = [matrix[i][1] for i in range(3)]

main_sum = sum(main_diagonal)
secondary_sum = sum(secondary_diagonal)
middle_sum = sum(middle_column)

print("main diagonale:", main_diagonal)
print("main sum:", main_sum)

print("secondary diagonale:", secondary_diagonal)
print("main sum:", secondary_sum)

print("Middle column:", middle_column)
print("Middle sum:", middle_sum)
```

Q10. List Transformation Challenge
data = [15, 5, 25, 10, 35, 20, 30]
Build a list of the three largest unique values ascending, then two smallest unique values descending. Expected: [25,
30, 35, 10, 5].

```python
data = [15,5,25,10,35,20,30]

# Sorted Data
data1 = sorted(set(data))
print(data1)

# List of Largest values
largest = data1[-3:]
print(largest)

# List of Smallest values
smallest = data1[:2]
print(smallest)

# Reverse List
smallest[::-1]

# Combine Lists
combine = largest + smallest
print("The Final Result is:", combine)
```

Q38. Order Calculation

order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

Calculate subtotals, grand total, and amount after 10% discount.

```python
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

grand_total = 0

print("--- Subtotals ---")
for item, details in order.items():
    subtotal = details["price"] * details["quantity"]
    grand_total += subtotal
    print(f"{item.capitalize()}: {details['quantity']} x {details['price']} = {subtotal}")

# Apply 10% discount
discount_rate = 0.10
discount_amount = grand_total * discount_rate
final_amount = grand_total - discount_amount

print("-" * 25)
print(f"Grand Total: {grand_total}")
print(f"Discount (10%): {discount_amount}")
print(f"Amount After Discount: {final_amount}")
```

Q39. Employee Performance

employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

Calculate each average, top performer, and highest SQL score.

```python
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

averages = {}
highest_sql = 0
top_sql_scorer = ""

print("--- Individual Averages ---")
for name, scores in employees.items():
    # Calculate average using sum of values divided by the number of subjects
    avg = sum(scores.values()) / len(scores)
    averages[name] = avg
    print(f"{name}: {avg:.2f}")
    
    # Track highest SQL score
    if scores["sql"] > highest_sql:
        highest_sql = scores["sql"]
        top_sql_scorer = name

# Find the top performer based on the highest average score
top_performer = max(averages, key=averages.get)

print("-" * 25)
print(f"Top Performer: {top_performer} (Average: {averages[top_performer]:.2f})")
print(f"Highest SQL Score: {highest_sql} (scored by {top_sql_scorer})")
```

Q40. Inventory Management

inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

Calculate total value, most valuable item by price x stock, and update stock after selling two laptops and five mice.

```python
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

total_inventory_value = 0
item_values = {}

print("--- Individual Item Total Values ---")
for prod_id, details in inventory.items():
    # Calculate value for each product (price * stock)
    item_value = details["price"] * details["stock"]
    total_inventory_value += item_value
    item_values[prod_id] = item_value
    print(f"{details['name']} ({prod_id}): {details['stock']} units @ {details['price']} = {item_value}")

# Find the most valuable item by price x stock
most_valuable_id = max(item_values, key=item_values.get)
most_valuable_name = inventory[most_valuable_id]["name"]

print("-" * 40)
print(f"Total Inventory Value: {total_inventory_value}")
print(f"Most Valuable Item: {most_valuable_name} ({most_valuable_id}) with a stock value of {item_values[most_valuable_id]}")

# Update stock after selling 2 laptops and 5 mice
inventory["P101"]["stock"] -= 2  # Sold 2 laptops
inventory["P102"]["stock"] -= 5  # Sold 5 mice

print("-" * 40)
print("Updated Inventory Stock Levels:")
for prod_id, details in inventory.items():
    print(f"{details['name']}: {details['stock']} units remaining")
```

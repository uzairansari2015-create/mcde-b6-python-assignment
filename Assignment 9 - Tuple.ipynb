Unpack first two and last into variables; put all remaining elements into a list.

```python
data = (100,200,300,400,500,600)

first, second, *middle, last = data # * in front of middle is the key. It tells python : "Collect all left over items"  

print("First:", first)
print("second:", second)
print("Middle:", middle)
print("last:", last)
```

Q17. Nested Tuple Access
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)
Extract Sara second score, Ali average, and Ahmed highest score.

```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

# Sara's second score
sara_second = students[1][1][1]

# Ali's average
ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)

# Ahmed's highest score
ahmed_highest = max(students[2][1])

print("Sara's second score:", sara_second)
print("Ali's average:", round(ali_average, 2))
print("Ahmed's highest score:", ahmed_highest)
```

Q19. Tuple Matrix

matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

Extract diagonal, calculate its product, and build a tuple of row sums.

```python
# Step 1: Extract diagonal elements
diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])

# Step 2: Calculate diagonal product
product = diagonal[0] * diagonal[1] * diagonal[2]

# Step 3: Calculate row sums
row_sums = (
    sum(matrix[0]),
    sum(matrix[1]),
    sum(matrix[2])
)

# Display results
print("Diagonal:", diagonal)
print("Diagonal Product:", product)
print("Row Sums:", row_sums)
```

Q20. Inventory Tuple Challenge

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

Calculate inventory value, find most expensive product, and build reversed tuple of names

```python
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

# Step 1: Calculate total inventory value
inventory_value = (
    records[0][1] * records[0][2] +
    records[1][1] * records[1][2] +
    records[2][1] * records[2][2]
)

# Step 2: Find the most expensive product
most_expensive = max(
    records,
    key=lambda x: x[1]
)

# Step 3: Build reversed tuple of product names
names = (
    records[0][0],
    records[1][0],
    records[2][0]
)

reversed_names = names[::-1]

# Display results
print("Inventory Value:", inventory_value)
print("Most Expensive Product:", most_expensive)
print("Reversed Product Names:", reversed_names)
```

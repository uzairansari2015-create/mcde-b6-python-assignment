## Q5. Notification Dispatcher | Beginner
Create EmailNotification, SMSNotification, and PushNotification classes. Each must implement send(message)
with output specific to its channel. Write a single function notify_all(notifications, message) that iterates through
different notification objects and invokes send(). Add a WhatsAppNotification class without changing the
dispatcher. Explain where polymorphism occurs.

```python
class EmailNotification:
    def send(self, message):
        print(f"Email Notification: {message}")

class SMSNotification:
    def send(self, message):
        print(f"SMS Notification: {message}")

class PushNotification:
    def send(self, message):
        print(f"Push Notification: {message}")

class WhatsAppNotification:
    def send(self, message):
        print(f"WhatsApp Notification: {message}")

# Dispathcher function to send notifications
def notify_all(notifications, message):
    for notification in notifications:
        notification.send(message)

# Creating Objects of EmailNotification, SMSNotification, and PushNotification classes
email = EmailNotification()
sms = SMSNotification()
push = PushNotification()
wapp = WhatsAppNotification()

# Creating a list of Objects
notification = [email, sms, push, wapp]

# Calling the dispatcher function to send notifications
notify_all(notification, "Your order has been shipped.")


    # class SMSNotification:
    #     def __init__(self, recipient, message):
    #         self.recipient = recipient
    #         self.message = message

    #     def send(self, message):
    #         print(f"Sending SMS to: {self.recipient} with message: {self.message}")

    # class PushNotification:
    #     def __init__(self, recipient, message):
    #         self.recipient = recipient
    #         self.message = message

    #     def send(self, message):
    #         print(f"Sending Push Notification to: {self.recipient} with message: {self.message}")

```

## Q6. Checkout Discount Calculator | Intermediate
Create RegularCustomer, PremiumCustomer, and VIPCustomer classes, each implementing
final_price(amount). Apply discounts of 5%, 10%, and 20%, respectively. Write print_receipt(customer, amount)
so that it works with any customer type. For a bill of Rs. 10,000, show the final amount for each type. Reject
negative bill amounts. Extension: Add GoldCustomer with a 15% discount without modifying print_receipt().

```python
class RegularCustomer:
    def final_price(self, amount):
        discount = amount * 0.05
        return amount - discount

class PremiumCustomer:
    def final_price(self, amount):
        discount = amount * 0.10
        return amount - discount

class VIPCustomer:
    def final_price(self, amount):
        discount = amount * 0.20
        return amount - discount

class GoldCustomer:
    def final_price(self, amount):
        discount = amount * 0.15
        return amount - discount

def print_receipt(customer, amount):
    if amount <= 0:
        print("Invalid amount. Please enter a positive value.")
        return
    
    final_amount = customer.final_price(amount)
    print(f"Original Amount: {amount}, Final Amount after discount: {final_amount}")
```

```python
# Creating objects of different customer types
regular = RegularCustomer()
premium = PremiumCustomer()
vip = VIPCustomer()
gold = GoldCustomer()

# Calling the print_receipt function for each customer type
print("\n- - Regular Customer Receipt - -")
print_receipt(regular, 1000)

print("\n - - Premium Customer Receipt - -")
print_receipt(premium, 1000)

print("\n - - VIP Customer Receipt - -")
print_receipt(vip, 1000)

print("\n - - Gold Customer Receipt - -")
print_receipt(gold, 1000)
```

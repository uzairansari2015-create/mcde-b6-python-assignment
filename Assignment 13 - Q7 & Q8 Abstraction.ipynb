## Q7. Shape Calculator | Beginner
Use ABC and @abstractmethod to define an abstract Shape class with area() and perimeter(). Implement both
methods in Circle and Rectangle. Accept dimensions through constructors and use pi = 3.14159. Test a circle
with radius 5 and a rectangle with width 4 and height 6. Attempt to instantiate Shape directly and explain the
outcome.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        pi = 3.14159
        return pi * self.radius ** 2

    def perimeter(self):
        pi = 3.14159
        return 2 * pi * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(Self):
        return 2 * (Self.length * Self.width)
    
# Creating Object
circle1 = Circle(5)
rectangle1 = Rectangle(4, 6)

# Display the area and perimeter of the Circle
print("\n- - - - - - CIRCLE - - - - - -")
print(f"The Area of the Circle is:", circle1.area())
print(f"The Perimeter of the Circle is:", circle1.perimeter())

# # Display the area and perimeter of the Rectangle
print("\n- - - - - - RECTANGLE - - - - - -")
print(f"The Area of the Rectangle is:", rectangle1.area())
print(f"The Perimeter of the Rectangle is:", rectangle1.perimeter())

try:
    shape1 = Shape() # This will pop up an error.
except TypeError as e:
    print("Error:", e)

# Error: Can't instantiate abstract class Shape without an implementation for abstract methods 'area', 'perimeter'
# Becasue the shape is an abstract class and the implementation of the shape method is not defined. Python does not allow to create an 
# abstract class object directly
```

## Q8. Payment Gateway | Intermediate
Define an abstract PaymentMethod class with abstract methods pay(amount) and refund(amount). Implement
CardPayment, WalletPayment, and BankTransferPayment. Each method must display the payment method,
transaction type, and amount. Validate that transaction amounts are positive. Write process_transaction(method,
amount) to call pay() for any supported payment method. Extension: Add a new payment method without
changing process_transaction()

```python
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass

class CardPayment(PaymentMethod):

    def pay(self, amount):
        if amount <= 0:
            print(f"Invalid Payment Amount!")
        else:
            print("Payment Method: CardPayment.")
            print("Transaction Type: Payment.")
            print("Amount: Rs.", amount)

    def refund(self, amount):
        return super().refund(amount)

class WalletPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print(f"Invalid Refund Amount!")
        else:
            print("Payment Method: WalletPayment.")
            print("Transaction Type: Refund.")
            print("Amount: Rs.", amount)

    def refund(self, amount):
        return super().refund(amount)

class BankTransferPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print(f"Invalid Refund Amount!")
        else:
            print("Payment Method: BankTransferPayment.")
            print("Transaction Type: Refund.")
            print("Amount: Rs.", amount)

    def refund(self, amount):
        return super().refund(amount)

def process_transaction(method, amount):
    method.pay(amount)
```

```python
# Creating Object

card = CardPayment()
wallet = WalletPayment()
bank = BankTransferPayment()

process_transaction(card, 1000)
process_transaction(wallet, 2000)
process_transaction(bank, 3000)

card.refund(500)
wallet.refund(700)
bank.refund(1000)

# Creating a New child class
class JazzcashPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print(f"Invalid Refund Amount!")
        else:
            print(f"Payment method: Jazz Cash")
            print(f"Transaction Type: Payment")
            print(f"Amount: Rs", amount)

    def refund(self, amount):
        return super().refund(amount)

# Creating Jazzcash Object
Jazzcash1 = JazzcashPayment()

process_transaction(Jazzcash1, 1000)
```
